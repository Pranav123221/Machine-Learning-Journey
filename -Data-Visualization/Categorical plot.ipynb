```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset("tips")
```

```python
df
```

```python
sns.countplot(x = df['sex'] , hue= df['smoker'])
```

```python
sns.barplot(x = df['sex'] , y= df['total_bill'])
```

```python
sns.boxplot(x = df['total_bill'] , y = df['sex'] ,hue=df['day'], palette='rainbow')
```

```python
sns.violinplot(x = df['total_bill'] , y = df['sex'] ,hue=df['day'], palette='rainbow')
```

```python
sns.stripplot(x = df['day'] , y = df['total_bill'] , data = df)
```

```python
sns.swarmplot(x = df['day'] , y = df['total_bill'] , data = df)
```

```python
## combining swarmplot and violin plot
```

```python
sns.violinplot(x = df['tip'] , y = df['day'] , data = df)
sns.swarmplot(x = df['tip'] , y = df['day'] , data = df)
```
